```python
import pandas as pd
import os

# Read the raw data you created locally
csv_path = os.path.join("..", "data", "raw", "iris.csv")
column_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']
df = pd.read_csv(csv_path, names=column_names)

# Display dataset details
print("--- Dataset Shape ---")
print(df.shape)
print("\n--- Missing Values Check ---")
print(df.isnull().sum())
print("\n--- Target Distribution ---")
print(df['species'].value_counts())
```

```python
# Look at the mean, min, max, and standard deviation of flower parts
df.describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Visualise how features group together by species
sns.set_theme(style="ticks")
sns.pairplot(df, hue="species", palette="bright")
plt.show()
```
