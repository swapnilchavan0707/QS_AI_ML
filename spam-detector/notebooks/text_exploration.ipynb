```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Read the local data
csv_path = os.path.join("..", "data", "raw", "spam.csv")
df = pd.read_csv(csv_path)

# Print metrics
print(f"Total entries loaded: {len(df)}")
print("\nLabel breakdown:")
print(df['label'].value_counts())

# Plot distribution histogram
sns.set_theme(style="whitegrid")
sns.countplot(x='label', data=df, palette='Reds')
plt.title("Distribution of Spam vs Ham Messages")
plt.show()
```

```python
# Calculate character lengths for both categories
df['char_count'] = df['text'].str.len()

plt.figure(figsize=(8, 4))
sns.histplot(data=df, x='char_count', hue='label', kde=True, bins=15, palette='Set1', multiple='stack')
plt.title("Message Character Length Distribution")
plt.xlabel("Character Count")
plt.ylabel("Frequency")
plt.show()
```
